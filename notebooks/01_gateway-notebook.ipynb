```python
from google import genai

gemini_client = genai.Client()  # reads GEMINI_API_KEY
g_resp = gemini_client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="Name three uses for a paperclip.",
)
print(g_resp.text)
print(g_resp.usage_metadata)
```

```python
from openai import OpenAI

groq_client = OpenAI(
    api_key=os.environ["GROQ_API_KEY"],
    base_url="https://api.groq.com/openai/v1",
)
```

```python
models_list = groq_client.models.list()
for model in models_list.data:
    print(model.id)
```

```python
q_resp = groq_client.chat.completions.create(
    model="openai/gpt-oss-120b",
    max_tokens=200,
    messages=[{"role": "user", "content": "Name three uses for a paperclip."}],
)
print(q_resp.choices[0].message.content)
print(q_resp.usage)
```
